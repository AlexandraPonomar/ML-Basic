### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
# ВАШ КОД ЗДЕСЬ
import numpy as np

m = np.zeros(10)
m[4] = 1
print('Исходный массив:', m)
m = np.reshape(m, (2, 5))
print('Массив после трансформации:', m)
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
# ВАШ КОД ЗДЕСЬ
import numpy as np

m = np.random.randint(10, 50, size=10)
m = np.flip(m)
iven_m = m[m % 2 == 0]
print(m)
print(iven_m)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
# ВАШ КОД ЗДЕСЬ
import numpy as np

m = np.random.randint(-100, 101, size=(4, 3, 2))
print('Трехмерный массив 4*3*2:', m)
print('Наименьшее число', np.min(m))
print('Наибольшее число', np.max(m))
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
# ВАШ КОД ЗДЕСЬ
m1 = np.random.randint(-100, 101, size=(6, 4))
m2 = np.random.randint(-100, 101, size=(4, 3))
print(m1 @ m2)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
# ВАШ КОД ЗДЕСЬ
import numpy as np

m = np.random.rand(7, 7)
mean_value = np.mean(m)
print('Среднее отклонение', mean_value)
stnd_value = np.std(m)
print('Стандартное отклонение', stnd_value)
min_m = np.min(m)
max_m = np.max(m)
normal_m = (m - min_m) / (max_m - min_m)
print('Исходный массив:', m)
print("Нормализованный массив", normal_m)
```
