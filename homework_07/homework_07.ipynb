### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
# ВАШ КОД ЗДЕСЬ
import numpy as np

m = np.random.rand(7, 7)
mean_value = np.mean(m)
print('Среднее отклонение', mean_value)
stnd_value = np.std(m)
print('Стандартное отклонение', stnd_value)
min_m = np.min(m)
max_m = np.max(m)
normal_m = (m - min_m) / (max_m - min_m)
print('Исходный массив:', m)
print("Нормализованный массив", normal_m)
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
df = sns.load_dataset('tips')
median_bill = df['total_bill'].median()
high_bill = df[df['total_bill'] > median_bill]
result = high_bill.groupby('sex')['tip'].mean()

print(result)
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
sns.scatterplot(
    data=df[df['sex']=='Male'],
    x='total_bill',
    y='tip',
    hue='smoker',   # hue - цвет точек зависит от категориальной переменной 'smoker'
)
plt.title('For male');
plt.show()
sns.scatterplot(
    data=df[df['sex']=='Female'],
    x='total_bill',
    y='tip',
    hue='smoker',   # hue - цвет точек зависит от категориальной переменной 'smoker'
)
plt.title('For Female');
plt.show()
```
