## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
# ВАШ КОД ЗДЕСЬ
user_str = input('Ведите текст для конвертации:')


def pascal_case(s):
    s = s.replace('_', ' ')
    s = s.title()
    s = s.replace(' ', '')
    return s


def snake_case(s):
    for i in range(len(s)):
        if s[i].isupper():
            first_part_s = s[:i]
            second_part = s[(i+1):]
            if i>0:
                s = first_part_s +  '_' + s[i].lower() + second_part
            else:
                s = s[i].lower() + second_part
    return s

if user_str.isalnum()==True and user_str.isalpha()==True:
    print(snake_case(user_str))
else:
    print(pascal_case(user_str))
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
# ВАШ КОД ЗДЕСЬ
num_input = int(input())
def sasha_simpl(num):
    i=2
    while i<num:
        if num%i==0:
            return False
        i+=1

    return True
print(sasha_simpl(num_input))
```

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
# ВАШ КОД ЗДЕСЬ
def print_table_base(table):
    print('\nID      | Имя          | Фамилия           | Возраст')
    print('-'*40)
    for usr_id in table:
        table_get = table.get(usr_id)

        print(usr_id ,' | ', table_get[0], ' | ', table_get[1], ' | ', table_get[2])


def usr_table():
    table = {}
    while True:
        usr_name = input('Введите имя:')
        if not usr_name:
            break
        usr_surname = input('Введите фамилию:')
        if not usr_surname:
            break
        usr_age = int(input('Введите возраст:'))
        if not usr_age:
            break
        usr_id = f"{int(input('Введите ID:')):08d}"
        if not usr_id:
            break
        usr_validate = True
        if usr_name.isalpha() and usr_surname.isalpha():
            usr_name = usr_name.capitalize()
            usr_surname = usr_surname.capitalize()
        else:
            usr_validate = False
            print('Ошибка: вы ввели неверное имя или фамилию')
        if not (18<=usr_age<=60):
            usr_validate = False
            print("Ошибка: возраст должен быть от 18 до 60 лет!")

        if usr_name not in table and usr_validate == True:
            table[usr_id] = (usr_name, usr_surname, usr_age)
        else:
            print('Ошибка: Пользователь не добавлен!')

    return table

table_output = usr_table()
print_table_base(table_output)
        
```
